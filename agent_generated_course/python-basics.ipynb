## What is Python?
Python is a high-level, interpreted programming language known for its simple syntax and readability. It's used for web development, data analysis, automation, AI, and much more. Python code is easy to write and understand, which makes it ideal for beginners and professionals alike.


**Variables** are names that refer to values stored in memory. You use variables to store data so you can work with it later.

**Data Types** describe the kind of data a variable can hold. In Python, data types are assigned automatically (dynamic typing). Common data types include:

int (integer numbers)
float (decimal numbers)
str (text, aka strings)
bool (True/False values)

```python
# Assigning a string value
x = "Hello World!"
print(x, type(x))  # Output: Hello World! <class 'str'>

**Step-by-Step Breakdown**: 
<br>
+ Assignment (=): Assigns a value to a variable name.
+ Type Inference:
+ Python figures out the type based on the assigned value—no need to declare a type.
+ type() Function:
+ Returns the type of the variable.
+ Printing: Displays both the value and its type.
+ Case Sensitivity: Variable names are case-sensitive (Height ≠ height).
+ Reassignment: You can change a variable’s type by assigning a new value:

````python
# Assigning an integer value
x = 10
print(x, type(x))  # Output: 10 <class 'int'>

# Reassigning a string value
x = "Now I'm a string!"
print(x, type(x))  # Output: Now I'm a string! <class 'str'>

# Reassigning a list value
x = [1, 2, 3]
print(x, type(x))  # Output: [1, 2, 3] <class 'list'>
````

**Type Hints (since Python 3.5+):**

You can add type hints for better clarity and tooling, but Python won’t enforce them at runtime.

````python
age: int = 20
age = "twenty"  # Still allowed, but type checkers will warn you
````

#### **User Input**

You can get user input in Python by using the input() function. It returns a string value.

```python
name = input("Enter your name: ")
print("Hello, " + name + "!")
```
input() always returns a string—even if the user types a number.

````python
# Ask the user for their name and age
name = input("What's your name? ")
age = input("How old are you? ")

print("Hello,", name + "! You are", age, "years old.")
````

**Step-by-Step Breakdown**

+ Prompt the user:
  + input("What's your name? ") displays the message and waits for user input.
  + The value the user types is stored in the variable name.
+ Ask for age:
  + input("How old are you? ") does the same.
  + The value is stored in age.
+ Print a response:
  + Combines the user’s inputs in a greeting.


**Common Pitfalls**
+ Type Conversion:
input() returns a string. If you want a number, you must convert:

````python
age = int(input("How old are you? "))  # Converts input to an integer
# If the user types something that isn’t a number, Python will raise a ValueError.
````

+ Error Handling:
Always handle possible exceptions:

````python
try:
    age = int(input("Enter your age: "))
except ValueError:
    print("Please enter a valid number!")
````

+ Whitespace:
User input may include trailing spaces. You can use .strip() to clean it:

````python
name = input("What's your name? ").strip()
````

**Best Practices and Modern Python Idioms**
+ Always validate and sanitize user input.
+ Use meaningful prompts to guide users.
+ For repeated input, use loops:

````python
while True:
    age = input("Enter your age: ")
    if age.isdigit():
        age = int(age)
        break
    print("Please enter a valid number.")
````

#### **Basic Math Operations**

Python supports all the standard arithmetic operations you’d expect, such as addition, subtraction, multiplication, and division. You can use these directly with numbers (integers and floats) and store results in variables.

Supported Operators:

+ Addition: +
+ Subtraction: -
+ Multiplication: *
+ Division: /
+ Integer Division: //
+ Modulus (remainder): %
+ Exponentiation: **

````python
# Basic math operations in Python

# Addition
a = 7 + 3         # 10

# Subtraction
b = 15 - 4        # 11

# Multiplication
c = 5 * 6         # 30

# Division (always returns a float)
d = 20 / 4        # 5.0

# Integer Division (drops remainder)
e = 20 // 3       # 6

# Modulus (returns remainder)
f = 20 % 3        # 2

# Exponentiation (power)
g = 2 ** 3        # 8

# Print results
print("Addition:", a)
print("Subtraction:", b)
print("Multiplication:", c)
print("Division:", d)
print("Integer Division:", e)
print("Modulus:", f)
print("Exponentiation:", g)
````

**Common Pitfalls**

+ Division always returns float: Even if dividing two integers, / returns a float.

````python
print(9 / 3)  # Output: 3.0 (not 3)
````

+ Integer Division: Use // if you want to drop decimals.
+ Modulus with negative numbers: The result follows the sign of the divisor

````python
print(-7 % 3)  # Output: 2
````

+ Operator Precedence: Multiplication, division, and modulus are done before addition/subtraction, unless you use parentheses.

````python
print(2 + 3 * 4)      # Output: 14
print((2 + 3) * 4)    # Output: 20
````

**Related Concepts & Advanced Variations**
+ Math Module: For advanced math, use Python’s math module:

````python
import math
print(math.sqrt(16))   # 4.0
print(math.pi)         # 3.141592653589793
print(math.pow(2, 3))   # 8.0
print(math.factorial(5))  # 120
````

+ Type Conversion: Mixing integers and floats results in a float.

````python
print(4 + 2.5)         # Output: 6.5
print(2.5 + 4)         # Output: 6.5
print(4 * 2.5)         # Output: 10.0
print(2.5 * 4)         # Output: 10.0
print(4 / 2.5)         # Output: 1.6
print(2.5 / 4)         # Output: 0.6
````

+ Rounding: Use round() to round numbers.

````python
print(round(3.14159, 2))  # Output: 3.14
````

#### **Control Flow**
Control flow lets your program choose what to do next based on conditions.

+ **if** checks if a condition is true and runs code if it is.
+ **elif** (“else if”) checks another condition if the first was false.
+ **else** runs if none of the previous conditions were true.
Think of it as asking questions:
“If it rains, take an umbrella. If it’s windy, wear a jacket. Otherwise, enjoy the sunshine!”

````python
# Simple control flow example

temperature = int(input("What is the temperature (°C)? "))

if temperature < 10:
    print("It's cold! Wear a coat.")
elif temperature < 20:
    print("It's cool. A sweater should be fine.")
else:
    print("It's warm! Enjoy the weather.")
````

**Step-by-Step Breakdown**

+ Get user input:

    + input() reads a string from the user.
    + int() converts it to an integer for comparison.

+ First condition:

    + If temperature < 10, print “It's cold! Wear a coat.”

+ Second condition:

    + If the first condition is false, check temperature < 20.
    + If true, print “It's cool. A sweater should be fine.”

+ Else:
    + If neither condition is true, print It's warm! Enjoy the weather.

**Common Pitfalls**

+ Indentation:
    + Python uses indentation (spaces) to define code blocks. All code inside an if, elif, or else must be indented.

+ Type Errors:
    + Comparing strings and numbers can cause errors. Always convert user input to the right type.

        ```python 
        age = int(input("Enter your age: "))
        ```
+ Multiple conditions:
    + Use logical operators and, or, not:

        ```python
        if age > 18 and age < 65:
            print("Adult")
        else:
            print("Not an adult")
        ```


**Related Concepts & Advanced Variations**

+ Nested Conditions:
  + You can put if statements inside each other.
    ```python
    if temperature > 0:
        if temperature < 10:
            print("Chilly!")
    ```

+ Ternary Operator:
    + Short “if” for simple assignments:
        ```python
        message = "Cold" if temperature < 10 else "Warm"
        ```

+ Match Statement (Python 3.10+):
    + For pattern matching (like switch-case):
        ```python
        match day_of_week:
            case "Monday":
                print("Start of week!")
            case "Friday":
                print("Almost weekend!")
            case _:
                print("Just another day.")
        ```

### **Loops**
Python has two main loop types:

+ **for loops:** Iterate over sequences (lists, strings, ranges, etc.).
+ **while loops:** Repeat as long as a condition is true.

Loops help you avoid repeating code manually and let your programs scale to handle any amount of data.

````python
# Print each item in a list using a for loop
fruits = ['apple', 'banana', 'cherry']

for fruit in fruits:
    print("I like", fruit)
````

````python
# Countdown using a while loop
count = 5

while count > 0:
    print("Counting down:", count)
    count -= 1  # Decrement count by 1

print("Blast off!")
````

**Step-by-Step Breakdown**

**For Loop**
+ Define a list called fruits.
+ The for loop picks each item (fruit) from the list, one at a time.
+ It prints a message for each fruit.

**While Loop**
+ Set count to 5.
+ The loop runs as long as count is greater than 0.
+ Inside the loop, print the current count and decrease it by 1.
+ When count becomes 0, the loop stops and prints "Blast off!"


**Common Pitfalls**
+ Infinite loops: A while loop with a condition that never becomes False will run forever.

    ```python
    while True:
        print("Infinite loop!")
    ```
+ Off-by-one errors: Be careful with ranges and loop boundaries.

    ```python
    for i in range(10):  # i goes from 0 to 9 (not 10)
        print(i)
    ```

**Related Concepts & Advanced Variations**
+ range() function: Generates a sequence of numbers.


````python
for i in range(1, 6):  # 1 to 5
    print(i)

print("-------------------------")

for i in range(1, 6, 2):  # 1 to 5, stepping by 2
    print(i)

print("-------------------------")

for i in range(5):
    print(i)
````

+ Enumerate: Get both index and value

````python
for index, fruit in enumerate(fruits):
    print(index, fruit)
````

+ Break/Continue: Control loop flow.

````python
for fruit in fruits:
    if fruit == 'banana':
        continue  # Skip banana
    print(fruit)
````

+ List Comprehension: Concise looping for lists.

````python
upper_fruits = [fruit.upper() for fruit in fruits]
print(upper_fruits)
````

**Best Practices and Modern Python Idioms**

1. Prefer for loops for iterating over sequences.
2. Use while loops when you don’t know how many times you'll repeat.
3. Avoid modifying collections while looping over them.
4. Use descriptive variable names.
5. Consider enumerate for index-value pairs.
6. Use list comprehensions for simple transformations.

**Functions**

A function is a named block of code that performs a specific task.

+ Functions help you avoid repeating code by letting you “call” them wherever you need.
+ They improve readability and maintainability.
+ Functions can accept arguments (inputs) and return values (outputs).

````python
# Define a function that greets a user
def greet(name):
    """Print a greeting message for the user."""
    print("Hello,", name + "!")

# Call the function
greet("Alice")
greet("Bob")

# Function with return value
def add(a, b):
    """Return the sum of two numbers."""
    return a + b

result = add(3, 5)
print("3 + 5 =", result)
````

**Step-by-Step Breakdown**

1. Define a function:
    + Use def followed by the function name and parentheses with parameters.
    + The indented code block is the function’s body.
2. Docstring (optional):
    + The triple-quoted string describes what the function does.
3. Call the function:
    + Use the function’s name followed by parentheses and arguments.
4. Return statement:
    + return sends a value back to the caller. You can assign this result to a variable.

**Common Pitfalls**
+ No return value: If you forget return, your function returns None by default.
+ Parameter mismatch: Calling a function with the wrong number/type of arguments causes errors.
+ Variable scope: Variables inside a function are local and not accessible outside.

    ```python
    def foo():
        x = 10  # x only exists inside foo
    print(x)  # NameError!
    ```


**Related Concepts & Advanced Variations**

+ Default parameters:

    ```python
    def greet(name="Guest"):
        print("Hello,", name)
    ```
+ Keyword arguments:

    ```python
        greet(name="Carol")
    ```

+ Variable number of arguments:

    ```python
    def add(*args):
        total = 0
        for num in args:
            total += num
        return total
    print(add(1, 2, 3))  # Output: 6
    ```
+ Lambda (anonymous) functions:

    ```python
    add = lambda a, b: a + b
    print(add(3, 5))  # Output: 8
    ```
+ Type hints (Python 3.5+):

    ```python
    def add(a: int, b: int) -> int:
        return a + b
    ```

**Best Practices and Modern Python Idioms**

+ Use clear, descriptive names for functions and arguments.
+ Write a docstring for every function (PEP 257 style).
+ Keep functions short and focused—do one thing well.
+ Use return values for data, and only use print for user interaction.
+ Prefer keyword arguments for clarity in function calls.
+ Avoid side effects unless necessary (pure functions are easier to test).

**Decorators** 

Are a powerful and elegant feature in Python that let you “wrap” or modify the behavior of functions or classes without changing their actual code. They’re widely used for logging, access control, timing, memoization, and more.

A **decorator** is a function that takes another function (or class) as an argument, and returns a new function (or class) with extra features.

+ Think of a decorator like putting a gift in a box: the box adds something (protection, wrapping) but the gift inside stays the same.
+ You use the @decorator_name syntax to apply a decorator to a function.

````python
def shout_decorator(func):
    def wrapper(*args, **kwargs):
        print("Before the function runs!")
        result = func(*args, **kwargs)
        print("After the function runs!")
        return result
    return wrapper

@shout_decorator
def greet(name):
    print(f"Hello, {name}!")

greet("Alice")
````

**Step-by-Step Breakdown**

1. Define the decorator function

    1. shout_decorator(func) receives a function as its argument.
    2. Inside, it defines wrapper, which calls the original func, adding extra behavior before and after.

2. Apply the decorator
    1. @shout_decorator above greet is the same as:
    2. greet = shout_decorator(greet)

3. Call the decorated function
    1. When you call `greet("Alice")`, you actually call the wrapper function.
